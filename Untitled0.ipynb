```python
import random
import time
from collections import deque
import matplotlib.pyplot as plt
import numpy as np

# ==========================================
# 1. LÓGICA DEL 8-PUZZLE
# ==========================================
GOAL = (1, 2, 3, 4, 5, 6, 7, 8, 0)

def get_neighbors(state):
    """Genera los movimientos válidos desde un estado dado."""
    neighbors = []
    idx = state.index(0)
    row, col = divmod(idx, 3)

    moves = []
    if row > 0: moves.append(-3)  # Arriba
    if row < 2: moves.append(3)   # Abajo
    if col > 0: moves.append(-1)  # Izquierda
    if col < 2: moves.append(1)   # Derecha

    for m in moves:
        new_state = list(state)
        new_state[idx], new_state[idx + m] = new_state[idx + m], new_state[idx]
        neighbors.append(tuple(new_state))
    return neighbors

def is_solvable(state):
    """Verifica si el puzzle tiene solución (inversiones pares)."""
    inv = 0
    s = [x for x in state if x != 0]
    for i in range(len(s)):
        for j in range(i + 1, len(s)):
            if s[i] > s[j]:
                inv += 1
    return inv % 2 == 0

def get_random_state(moves=10):
    """Genera un estado inicial aleatorio con N movimientos desde la meta."""
    state = GOAL
    for _ in range(moves):
        state = random.choice(get_neighbors(state))
    return state

# ==========================================
# 2. ALGORITMOS CON MÉTRICAS COMPLETAS
# ==========================================

def bfs(start_state):
    """BFS con métricas de rendimiento."""
    start_time = time.time()
    queue = deque([start_state])
    visited = {start_state: None}  # state: parent
    nodes_expanded = 0
    max_frontier = 1

    while queue:
        max_frontier = max(max_frontier, len(queue))
        current = queue.popleft()
        nodes_expanded += 1

        if current == GOAL:
            # Reconstruir camino
            path = []
            while current is not None:
                path.append(current)
                current = visited[current]
            return {
                "time": time.time() - start_time,
                "nodes": nodes_expanded,
                "path": path[::-1],  # Invertir para ir del inicio a la meta
                "memory": len(visited) + max_frontier
            }

        for neighbor in get_neighbors(current):
            if neighbor not in visited:
                visited[neighbor] = current
                queue.append(neighbor)
    return None

def dfs(start_state):
    """DFS con métricas de rendimiento."""
    start_time = time.time()
    stack = [start_state]
    visited = {start_state: None}
    nodes_expanded = 0
    max_frontier = 1

    while stack:
        max_frontier = max(max_frontier, len(stack))
        current = stack.pop()
        nodes_expanded += 1

        if current == GOAL:
            path = []
            while current is not None:
                path.append(current)
                current = visited[current]
            return {
                "time": time.time() - start_time,
                "nodes": nodes_expanded,
                "path": path[::-1],
                "memory": len(visited) + max_frontier
            }

        for neighbor in get_neighbors(current):
            if neighbor not in visited:
                visited[neighbor] = current
                stack.append(neighbor)
    return None

# ==========================================
# 3. VISUALIZACIÓN DE TABLEROS
# ==========================================

def draw_board(ax, state, title=""):
    """Dibuja un tablero 3x3 en un eje de matplotlib."""
    if state is None:
        ax.axis('off')
        ax.text(0.5, 0.5, '...', fontsize=30, ha='center', va='center')
        return

    grid = np.array(state).reshape(3, 3)
    ax.matshow(np.zeros((3, 3)), cmap='Blues', alpha=0.1)

    for i in range(3):
        for j in range(3):
            val = grid[i, j]
            if val != 0:
                rect = plt.Rectangle((j - 0.5, i - 0.5), 1, 1,
                                     facecolor='lightgray', edgecolor='black', linewidth=2)
                ax.add_patch(rect)
                ax.text(j, i, str(val), va='center', ha='center',
                        fontsize=20, fontweight='bold', color='black')
            else:
                rect = plt.Rectangle((j - 0.5, i - 0.5), 1, 1,
                                     facecolor='white', edgecolor='gray', linewidth=1)
                ax.add_patch(rect)

    ax.set_xticks([])
    ax.set_yticks([])
    ax.set_title(title, pad=10, fontsize=12)

def plot_game_path(path, algo_name, max_displays=8):
    """
    Dibuja la secuencia de movimientos. Si el camino es muy largo,
    muestra solo el inicio y el final.
    """
    if path is None:
        print(f"No se encontró camino con {algo_name}")
        return

    if len(path) > max_displays:
        # Tomar primeros 4, separador, y últimos 3
        display_path = path[:4] + [None] + path[-3:]
        indices = list(range(4)) + [None] + list(range(len(path) - 3, len(path)))
    else:
        display_path = path
        indices = list(range(len(path)))

    fig, axes = plt.subplots(1, len(display_path), figsize=(3.5 * len(display_path), 4))
    if len(display_path) == 1:
        axes = [axes]

    fig.suptitle(f'Solución usando {algo_name} (Total: {len(path)-1} pasos)',
                 fontsize=18, y=1.15)

    for ax, state, step_idx in zip(axes, display_path, indices):
        if state is None:
            draw_board(ax, None)
        else:
            if step_idx == 0:
                title = "Estado Inicial"
            elif step_idx == len(path) - 1:
                title = "¡META!"
            else:
                title = f"Paso {step_idx}"
            draw_board(ax, state, title)

    plt.tight_layout()
    plt.show()

# ==========================================
# 4. GRÁFICAS DE COMPLEJIDAD
# ==========================================

def plot_comparison_metrics(bfs_metrics, dfs_metrics):
    """Genera gráficas de comparación BFS vs DFS."""
    fig, axs = plt.subplots(2, 2, figsize=(12, 10))
    fig.suptitle('Comparación BFS vs DFS (Promedio de 10 partidas)', fontsize=16)
    labels = ['BFS', 'DFS']

    def plot_bar(ax, data_bfs, data_dfs, title, ylabel, color1='skyblue', color2='salmon'):
        means = [np.mean(data_bfs), np.mean(data_dfs)]
        ax.bar(labels, means, color=[color1, color2])
        ax.set_title(title)
        ax.set_ylabel(ylabel)
        for i, v in enumerate(means):
            ax.text(i, v + (v * 0.02), str(round(v, 2)),
                    ha='center', va='bottom', fontweight='bold')

    plot_bar(axs[0, 0], bfs_metrics["time"], dfs_metrics["time"],
             'Costo Temporal Real (Tiempo CPU)', 'Segundos')
    plot_bar(axs[0, 1], bfs_metrics["nodes"], dfs_metrics["nodes"],
             'Complejidad Temporal (Nodos Expandidos)', 'Cantidad de Nodos')
    plot_bar(axs[1, 0], bfs_metrics["path_len"], dfs_metrics["path_len"],
             'Calidad de la Solución (Pasos)', 'Movimientos hasta la meta')
    plot_bar(axs[1, 1], bfs_metrics["memory"], dfs_metrics["memory"],
             'Complejidad Espacial (Nodos Guardados)', 'Nodos en RAM')

    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.savefig('graficas_comparacion_algoritmos.png', dpi=300)
    print("Gráfica guardada exitosamente como 'graficas_comparacion_algoritmos.png'")
    plt.show()

# ==========================================
# 5. EJECUCIÓN PRINCIPAL
# ==========================================

def run_full_simulation():
    """Ejecuta la simulación completa con métricas y visualización."""

    print("=" * 60)
    print("SIMULACIÓN DE 8-PUZZLE CON BFS Y DFS")
    print("(Estados iniciales completamente aleatorios)")
    print("=" * 60)

    # --- Generar estado aleatorio para visualización (8 movimientos) ---
    print("\n1. VISUALIZACIÓN DE CAMINOS (con un estado aleatorio)")
    print("-" * 40)
    estado_prueba = get_random_state(moves=8)   # <--- AHORA ALEATORIO
    print(f"Estado inicial generado aleatoriamente: {estado_prueba}")
    print(f"¿Es solucionable? {is_solvable(estado_prueba)}")

    print("\nCalculando camino con BFS...")
    bfs_result = bfs(estado_prueba)
    if bfs_result:
        print(f"BFS encontró solución en {len(bfs_result['path'])-1} pasos, "
              f"expandiendo {bfs_result['nodes']} nodos en {bfs_result['time']:.4f}s")
        plot_game_path(bfs_result['path'], "BFS")

    print("\nCalculando camino con DFS...")
    dfs_result = dfs(estado_prueba)
    if dfs_result:
        print(f"DFS encontró solución en {len(dfs_result['path'])-1} pasos, "
              f"expandiendo {dfs_result['nodes']} nodos en {dfs_result['time']:.4f}s")
        plot_game_path(dfs_result['path'], "DFS")

    # --- Simulación de 10 partidas con estados aleatorios ---
    print("\n2. ANÁLISIS DE COMPLEJIDAD")
    print("-" * 40)

    trials = 10
    bfs_metrics = {"time": [], "nodes": [], "path_len": [], "memory": []}
    dfs_metrics = {"time": [], "nodes": [], "path_len": [], "memory": []}

    for i in range(trials):
        # Cada iteración genera un nuevo estado aleatorio con 12 movimientos
        initial = get_random_state(moves=12)
        print(f"\nPartida {i+1}: Estado inicial = {initial}")

        # BFS
        print(f"  Ejecutando BFS...")
        res_bfs = bfs(initial)
        bfs_metrics["time"].append(res_bfs["time"])
        bfs_metrics["nodes"].append(res_bfs["nodes"])
        bfs_metrics["path_len"].append(len(res_bfs["path"]) - 1)
        bfs_metrics["memory"].append(res_bfs["memory"])
        print(f"    BFS: {len(res_bfs['path'])-1} pasos, {res_bfs['nodes']} nodos, {res_bfs['time']:.4f}s")

        # DFS
        print(f"  Ejecutando DFS...")
        res_dfs = dfs(initial)
        dfs_metrics["time"].append(res_dfs["time"])
        dfs_metrics["nodes"].append(res_dfs["nodes"])
        dfs_metrics["path_len"].append(len(res_dfs["path"]) - 1)
        dfs_metrics["memory"].append(res_dfs["memory"])
        print(f"    DFS: {len(res_dfs['path'])-1} pasos, {res_dfs['nodes']} nodos, {res_dfs['time']:.4f}s")

    # Mostrar promedios
    print("\n" + "=" * 40)
    print("RESUMEN DE RESULTADOS")
    print("=" * 40)
    print(f"BFS - Tiempo: {np.mean(bfs_metrics['time']):.4f}s, "
          f"Nodos: {np.mean(bfs_metrics['nodes']):.1f}, "
          f"Pasos: {np.mean(bfs_metrics['path_len']):.1f}, "
          f"Memoria: {np.mean(bfs_metrics['memory']):.1f}")
    print(f"DFS - Tiempo: {np.mean(dfs_metrics['time']):.4f}s, "
          f"Nodos: {np.mean(dfs_metrics['nodes']):.1f}, "
          f"Pasos: {np.mean(dfs_metrics['path_len']):.1f}, "
          f"Memoria: {np.mean(dfs_metrics['memory']):.1f}")

    # Generar gráficas de comparación
    print("\n3. GENERANDO GRÁFICAS DE COMPARACIÓN")
    print("-" * 40)
    plot_comparison_metrics(bfs_metrics, dfs_metrics)

    print("\n" + "=" * 60)
    print("SIMULACIÓN COMPLETADA")
    print("=" * 60)

# Ejecutar la simulación
if __name__ == "__main__":
    run_full_simulation()
```

# PUZZLE 3x3

En este problema se utiliza un puzzle de 3x3 donde el espacio vacío
se representa con el número 0.

Se comparan los algoritmos BFS y DFS para encontrar una solución.

Se analizará:
- Tiempo de ejecución.
- Nodos explorados.
- Profundidad de la solución.
- Árbol de búsqueda.

```python
from collections import deque
import time
import matplotlib.pyplot as plt

inicio_puzzle = (1,2,3,5,0,6,4,7,8)
meta_puzzle = (1,2,3,4,5,6,7,8,0)

def movimientos(estado):
    resultado = []
    pos = estado.index(0)
    fila, col = pos // 3, pos % 3

    for df, dc in [(-1,0),(1,0),(0,-1),(0,1)]:
        nf, nc = fila + df, col + dc

        if 0 <= nf < 3 and 0 <= nc < 3:
            nueva = list(estado)
            nueva[pos], nueva[nf*3+nc] = nueva[nf*3+nc], nueva[pos]
            resultado.append(tuple(nueva))

    return resultado
```

BFS y DFS


```python
def bfs():
    cola = deque([(inicio_puzzle, [])])
    visitados = {inicio_puzzle}
    nodos = 0

    while cola:
        estado, camino = cola.popleft()
        nodos += 1

        if estado == meta_puzzle:
            return camino, nodos

        for nuevo in movimientos(estado):
            if nuevo not in visitados:
                visitados.add(nuevo)
                cola.append((nuevo, camino + [nuevo]))

def dfs():
    pila = [(inicio_puzzle, [])]
    visitados = {inicio_puzzle}
    nodos = 0

    while pila:
        estado, camino = pila.pop()
        nodos += 1

        if estado == meta_puzzle:
            return camino, nodos

        for nuevo in movimientos(estado):
            if nuevo not in visitados:
                visitados.add(nuevo)
                pila.append((nuevo, camino + [nuevo]))
```

Resultados

```python
inicio = time.perf_counter()
sol_bfs, nodos_bfs = bfs()
tiempo_bfs = time.perf_counter() - inicio

inicio = time.perf_counter()
sol_dfs, nodos_dfs = dfs()
tiempo_dfs = time.perf_counter() - inicio

print("BFS")
print("Nodos:", nodos_bfs)
print("Movimientos:", len(sol_bfs))
print("Tiempo:", tiempo_bfs)

print("\nDFS")
print("Nodos:", nodos_dfs)
print("Movimientos:", len(sol_dfs))
print("Tiempo:", tiempo_dfs)
```

Arbol

Cada nodo representa un estado del puzzle y cada rama representa un movimiento posible.

Graficas

```python
algoritmos = ["BFS", "DFS"]

plt.bar(algoritmos, [nodos_bfs, nodos_dfs])
plt.title("Nodos explorados")
plt.ylabel("Nodos")
plt.show()

plt.bar(algoritmos, [tiempo_bfs, tiempo_dfs])
plt.title("Tiempo de ejecución")
plt.ylabel("Segundos")
plt.show()
```



## Análisis

BFS explora el problema por niveles, mientras que DFS profundiza en
un camino antes de regresar.

Los resultados muestran diferencias en el tiempo y en la cantidad de
nodos explorados. BFS normalmente encuentra una solución con menor
cantidad de movimientos, mientras que DFS puede utilizar otra ruta.

# N-REINAS

Se comparan BFS y DFS para encontrar una solución al problema de N-Reinas.
El objetivo es colocar 8 reinas sin que ninguna pueda atacar a otra.

```python
from collections import deque
import time

N = 8

# Revisar si podemos colocar una reina
def es_seguro(tablero, fila, columna):

    for fila_anterior in range(fila):
        columna_anterior = tablero[fila_anterior]

        # Misma columna
        if columna_anterior == columna:
            return False

        # Misma diagonal
        if abs(columna_anterior - columna) == abs(fila_anterior - fila):
            return False

    return True


# -------------------------
# DFS
# -------------------------

def n_reinas_dfs(N):

    pila = [()]
    nodos = 0

    while pila:

        tablero = pila.pop()
        nodos += 1

        fila = len(tablero)

        # Solución encontrada
        if fila == N:
            return tablero, nodos

        # Intentamos colocar una reina
        for columna in range(N - 1, -1, -1):

            if es_seguro(tablero, fila, columna):
                nuevo = tablero + (columna,)
                pila.append(nuevo)

    return None, nodos


# -------------------------
# BFS
# -------------------------

def n_reinas_bfs(N):

    cola = deque([()])
    nodos = 0

    while cola:

        tablero = cola.popleft()
        nodos += 1

        fila = len(tablero)

        # Solución encontrada
        if fila == N:
            return tablero, nodos

        for columna in range(N):

            if es_seguro(tablero, fila, columna):
                nuevo = tablero + (columna,)
                cola.append(nuevo)

    return None, nodos


# Ejecutar DFS
inicio = time.perf_counter()
solucion_dfs, nodos_dfs = n_reinas_dfs(N)
tiempo_dfs = time.perf_counter() - inicio


# Ejecutar BFS
inicio = time.perf_counter()
solucion_bfs, nodos_bfs = n_reinas_bfs(N)
tiempo_bfs = time.perf_counter() - inicio


print("===== N REINAS =====")

print("\nDFS")
print("Solución:", solucion_dfs)
print("Nodos explorados:", nodos_dfs)
print("Tiempo:", tiempo_dfs, "segundos")

print("\nBFS")
print("Solución:", solucion_bfs)
print("Nodos explorados:", nodos_bfs)
print("Tiempo:", tiempo_bfs, "segundos")
```

```python
def mostrar_tablero(solucion):

    for fila in range(N):

        for columna in range(N):

            if solucion[fila] == columna:
                print("♛", end=" ")
            else:
                print(".", end=" ")

        print()


print("\nTablero encontrado por DFS:")
mostrar_tablero(solucion_dfs)

print("\nTablero encontrado por BFS:")
mostrar_tablero(solucion_bfs)
```

#Bfs Y Dfs

```python
def bfs_reinas():
    cola = deque([()])
    nodos = 0

    while cola:
        tablero = cola.popleft()
        nodos += 1

        if len(tablero) == N:
            return tablero, nodos

        fila = len(tablero)

        for col in range(N):
            if seguro(tablero, fila, col):
                cola.append(tablero + (col,))

def dfs_reinas():
    pila = [()]
    nodos = 0

    while pila:
        tablero = pila.pop()
        nodos += 1

        if len(tablero) == N:
            return tablero, nodos

        fila = len(tablero)

        for col in range(N - 1, -1, -1):
            if seguro(tablero, fila, col):
                pila.append(tablero + (col,))
```


#Resultados

```python
inicio = time.perf_counter()
sol_bfs, nodos_bfs_reinas = bfs_reinas()
tiempo_bfs_reinas = time.perf_counter() - inicio

inicio = time.perf_counter()
sol_dfs, nodos_dfs_reinas = dfs_reinas()
tiempo_dfs_reinas = time.perf_counter() - inicio

print("N-REINAS")

print("\nBFS")
print("Solución:", sol_bfs)
print("Nodos:", nodos_bfs_reinas)
print("Tiempo:", tiempo_bfs_reinas)

print("\nDFS")
print("Solución:", sol_dfs)
print("Nodos:", nodos_dfs_reinas)
print("Tiempo:", tiempo_dfs_reinas)
```

# Problema de la Mochila

Se comparan BFS y DFS para seleccionar objetos buscando obtener
el mayor valor posible sin superar la capacidad de la mochila.

#Datos


```python
objetos = [
    (2, 12),
    (1, 10),
    (3, 20),
    (2, 15),
    (4, 25)
]

capacidad = 7
```

#BFS Y DFS

```python
def mochila_bfs():
    cola = deque([(0, 0, 0)])
    mejor_valor = 0
    nodos = 0

    while cola:
        i, peso, valor = cola.popleft()
        nodos += 1

        if peso > capacidad:
            continue

        mejor_valor = max(mejor_valor, valor)

        if i < len(objetos):
            p, v = objetos[i]

            # No tomar el objeto
            cola.append((i + 1, peso, valor))

            # Tomar el objeto
            cola.append((i + 1, peso + p, valor + v))

    return mejor_valor, nodos


def mochila_dfs():
    pila = [(0, 0, 0)]
    mejor_valor = 0
    nodos = 0

    while pila:
        i, peso, valor = pila.pop()
        nodos += 1

        if peso > capacidad:
            continue

        mejor_valor = max(mejor_valor, valor)

        if i < len(objetos):
            p, v = objetos[i]

            pila.append((i + 1, peso, valor))
            pila.append((i + 1, peso + p, valor + v))

    return mejor_valor, nodos
```

#Resultados

```python
inicio = time.perf_counter()
valor_bfs, nodos_bfs_mochila = mochila_bfs()
tiempo_bfs_mochila = time.perf_counter() - inicio

inicio = time.perf_counter()
valor_dfs, nodos_dfs_mochila = mochila_dfs()
tiempo_dfs_mochila = time.perf_counter() - inicio

print("MOCHILA")

print("\nBFS")
print("Mejor valor:", valor_bfs)
print("Nodos:", nodos_bfs_mochila)
print("Tiempo:", tiempo_bfs_mochila)

print("\nDFS")
print("Mejor valor:", valor_dfs)
print("Nodos:", nodos_dfs_mochila)
print("Tiempo:", tiempo_dfs_mochila)
```

## Árbol de búsqueda

Cada nivel representa un objeto. En cada nivel existen dos opciones:
tomar el objeto o no tomarlo.
